```python
import pandas as pd

# ==========================
# HIER EINGEBEN
# ==========================
PLZ = "3012"
WOHNFLAECHE = 75
# ==========================


kurven = pd.read_csv(
    "mietmodell_kurven_30_180.csv",
    dtype={"plz": str}
)

kurven["plz"] = kurven["plz"].str.zfill(4)

r = kurven[
    (kurven["plz"] == PLZ.zfill(4)) &
    (kurven["area_m2"] == WOHNFLAECHE)
]

if r.empty:
    print("Keine Schätzung gefunden.")

else:
    r = r.iloc[0]

    q20 = r["q20_chf_m2_year"]
    q50 = r["q50_chf_m2_year"]
    q80 = r["q80_chf_m2_year"]

    m20 = q20 * WOHNFLAECHE / 12
    m50 = q50 * WOHNFLAECHE / 12
    m80 = q80 * WOHNFLAECHE / 12

    print(f"PLZ {PLZ} | {WOHNFLAECHE} m²")
    print()
    print(f"Q20:    {q20:,.0f} CHF/m²/Jahr  →  {m20:,.0f} CHF/Monat")
    print(f"Median: {q50:,.0f} CHF/m²/Jahr  →  {m50:,.0f} CHF/Monat")
    print(f"Q80:    {q80:,.0f} CHF/m²/Jahr  →  {m80:,.0f} CHF/Monat")
    print()
    print(f"Lokale Datendichte: {int(r['local_density'])}")
    print(f"Nachbargewicht: {r['neighbor_weight']:.1%}")
```
